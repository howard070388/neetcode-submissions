```python
import transformers
import peft
import datasets
import accelerate

from transformers import AutoTokenizer, AutoModelForCausalLM
from datasets import dataset
from peft import LoraConfig, get_peft_model
```

```python
model_name = "Qwen/Qwen2.5-0.5B-Instruct"

model = AutoModelForCausalLM.from_pretrained(model_name)
tokenizer = AutoTokenizer.from_pretrained(model_name)

prompt = "What is LoRA."
messages = [
    {"role": "system", "content": "You are Qwen, created by Alibaba Cloud. You are a helpful assistant."},
    {"role": "user", "content": prompt}
]
text = tokenizer.apply_chat_template(
    messages,
    tokenize=False,
    add_generation_prompt=True
)

inputs = tokenizer(text, return_tensors="pt")

outputs = model.generate(
    **inputs,
    max_new_tokens=80
)

result = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print(result)
```

```python
def tokenize_for_training(example):
    # 只有 user 的部分
    prompt_messages = [
        {
            "role": "user",
            "content": example["instruction"]
        }
    ]

    # 完整的 user + assistant
    full_messages = [
        {
            "role": "user",
            "content": example["instruction"]
        },
        {
            "role": "assistant",
            "content": example["response"]
        }
    ]

    # 產生「user 問完，準備輪到 assistant 回答」的文字
    prompt_text = tokenizer.apply_chat_template(
        prompt_messages,
        tokenize=False,
        add_generation_prompt=True
    )

    # 產生完整的 user + assistant 文字
    full_text = tokenizer.apply_chat_template(
        full_messages,
        tokenize=False
    )

    # 轉成 token
    prompt_ids = tokenizer(
        prompt_text
    )["input_ids"]

    full_ids = tokenizer(
        full_text
    )["input_ids"]

    # 一開始先複製完整答案
    labels = full_ids.copy()

    # user 前面的部分全部改成 -100
    # 代表這些位置不計算 loss
    labels[:len(prompt_ids)] = [-100] * len(prompt_ids)

    return {
        "input_ids": full_ids,
        "attention_mask": [1] * len(full_ids),
        "labels": labels
    }
dataset = Dataset.from_list(train_data)

# 對整個 dataset 的每一筆資料執行
tokenized_dataset = dataset.map(tokenize_for_training)

# 檢查第一筆
print(tokenized_dataset[0]["labels"])
```

```python
lora_config = LoraConfig(
    r=8,
    lora_alpha=16,
    target_modules=["q_proj", "v_proj"],
    lora_dropout=0.1,
    bias="none",
    task_type="CAUSAL_LM"
)

lora_model = get_peft_model(model, lora_config)

lora_model.print_trainable_parameters()
```

```python
from transformers import TrainingArguments, Trainer, DataCollatorForLanguageModeling, DataCollatorForSeq2Seq

training_args = TrainingArguments(
    output_dir="./qwen_lora_test",
    num_train_epochs=5,
    per_device_train_batch_size=1,
    learning_rate=2e-4,
    logging_steps=1,
    save_strategy="no",
    report_to="none"
)

data_collator = DataCollatorForSeq2Seq(
    tokenizer=tokenizer,
    model=lora_model,
    padding=True,
    label_pad_token_id=-100
)
trainer = Trainer(
    model=lora_model,
    args=training_args,
    train_dataset=tokenized_dataset,
    data_collator=data_collator
)

trainer.train()
```
